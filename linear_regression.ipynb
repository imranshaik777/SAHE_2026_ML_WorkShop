```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
sns.set_style(style='whitegrid')
```

```python
##Step2: loading(autompg dataset) and cleaning the data 
data = fetch_ucirepo(id=9)
data
```

```python
x = data.data.features.copy()##it returns all numerical features
y = data.data.targets['mpg']
```

```python
x.info()
x.describe()
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python
x['horsepower']=pd.to_numeric(x['horsepower'],errors='coerce')
x['origin']=x['origin'].astype(str)
```

```python
x.info()
```

```python
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features =['origin']
x=x[numerical_features + categorical_features]
x.info()
x.isnull().sum()
```

```python
y = data.data.targets['mpg']
```

```python
x.shape
```

```python
#step3:Quick Exploratory data analysis (seaborn)
df=x.copy()
df['mpg']=y
df
```

```python
fig,axes = plt.subplots(1,3,figsize=(16,4))
sns.histplot(df['mpg'],kde=True,ax=axes[0],color='brown')
axes[0].set_title("MPG Distribution")
sns.scatterplot(x=df['weight'],y=df['mpg'],ax=axes[1],color='red')
axes[1].set_title("Weight vs MPG Distibution")
sns.scatterplot(x=df['horsepower'],y=df['mpg'],ax=axes[2],color='skyblue')
axes[2].set_title("HorsePower vs MPG DIstribution")
```

```python
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,cmap='YlGn',center=0)
plt.title("correlation matrix")
plt.tight_layout()
plt.show()
```

```python
#step4 dividing the data into train test split 
#print(x)
#print(y)
#split rules --> 70-30,75-25,80-20
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('\nTrain Test Split')
print("Training data",len(x_train))
print("Testing data",len(x_test))
#print(x_train)
```

```python
#step5 : creating an evaluation function for baseline model
def evaluate(model_name,actual,predicted):
    """Return MAE,MSE,RMSE,R2-Score"""
    mse=mean_squared_error(actual,predicted)
    mae = mean_absolute_error(actual,predicted)
    return{'Model_Name':model_name,
    'MAE':mean_absolute_error,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-Score':r2_score(actual,predicted)}
results=[]
```

```python
#step6 : Baseline model --> it predicts the training set for every values 
baseline_predictions=np.full(len(y_test),y_train.mean())
#baseline_predictions
#in above case  we are keeeping a baseline prediction
results.append(evaluate("Mean baseline",y_test,baseline_predictions))
#results
#in above case we are creating the baseline predicting data
```

```python
#step7: We apply Linear Regression
simple_model = LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions = simple_model.predict(x_test[['weight']])
#print(simple_model_predictions)
results.append(evaluate("Simple Linear Regressions:Weight",y_test,simple_model_predictions))
print(len(results))
```

```python
print(f"Simple Linear Regression Equation")
print(f"Predicted MPG = {simple_model.intercept_}+{simple_model.coef_[0]}*weight")
```
